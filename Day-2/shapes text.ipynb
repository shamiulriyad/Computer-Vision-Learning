```python
import cv2
import numpy as np

image = np.zeros((400, 400, 3), dtype=np.uint8)
print('image shape:', image.shape)

image[:] = (255, 0, 0)  # blue color in BGR format
cv2.line(image, (0, 0), (400, 400), (0, 255, 0), 5)  # green diagonal line
cv2.rectangle(image, (50, 50), (350, 350), (0, 0, 255), 3)  # red rectangle
cv2.circle(image, (200, 200), 100, (255, 255, 0), -1)  # cyan circle
cv2.putText(image, 'OpenCV', (100, 200), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)  # white text

cv2.imshow('output', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
