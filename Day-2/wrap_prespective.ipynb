```python
import cv2
import numpy as np  

width, height = 400, 400

image =cv2.imread('data/p2.jpg')

pts1 = np.float32([[752, 118], [1120, 265], [540, 668], [871, 830]])
pts2= np.float32([[0, 0], [width, 0], [50, height], [width-50, height]])

metrix = cv2.getPerspectiveTransform(pts1, pts2)
image_out = cv2.warpPerspective(image, metrix, (width, height))
cv2.imshow('output', image)
cv2.imshow('output', image_out)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
