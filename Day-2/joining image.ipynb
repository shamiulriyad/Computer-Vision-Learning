```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = 'data/images.jpg'
image = cv2.imread(image_path)

if image is None:
    raise FileNotFoundError(f"Could not load image: {image_path}")

image_horizontal = np.hstack((image, image))
image_vertical = np.vstack((image, image))

# OpenCV loads images as BGR, matplotlib expects RGB
plt.figure(figsize=(10, 4))
plt.imshow(cv2.cvtColor(image_horizontal, cv2.COLOR_BGR2RGB))
plt.title("Horizontal")
plt.axis("off")
plt.show()

plt.figure(figsize=(5, 8))
plt.imshow(cv2.cvtColor(image_vertical, cv2.COLOR_BGR2RGB))
plt.title("Vertical")
plt.axis("off")
plt.show()
```
